```python
import requests
import json
```

```python
endpoint = "https://<your-text-analytics-endpoint>.cognitiveservices.azure.com/"
api_key = "<your-text-analytics-api-key>"
```

```python
endpoint
```

```python
api_key
```

```python
sentiment_url = endpoint + "text/analytics/v3.1/sentiment"
```

```python
reviews = [
    "The movie was fantastic! I loved the storyline and the acting was superb.",
    "I did not enjoy the movie. The plot was too predictable and the characters were flat.",
    "An average movie. It had some good moments, but overall it was quite boring.",
]
```

```python
documents = {"documents": [{"id": str(i+1), "language": "en", "text": review} for i, review in enumerate(reviews)]}
```

```python
documents
```

```python
headers = {
    "Ocp-Apim-Subscription-Key": api_key,
    "Content-Type": "application/json"
}
```

```python
headers
```

```python
response = requests.post(sentiment_url, headers=headers, json=documents)
```

```python
if response.status_code == 200:
    sentiment_results = response.json()
    for document in sentiment_results['documents']:
        print(f"Review {document['id']}:")
        print(f"  Sentiment: {document['sentiment']}")
        print(f"  Positive Score: {document['confidenceScores']['positive']:.2f}")
        print(f"  Neutral Score: {document['confidenceScores']['neutral']:.2f}")
        print(f"  Negative Score: {document['confidenceScores']['negative']:.2f}")
        print()
else:
    print(f"Error {response.status_code}: {response.text}")
```
